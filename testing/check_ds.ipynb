```python
import json

dataset_path = "../samples/TreeTreeBudetDirka.json"

with open(dataset_path, encoding="utf-8") as f:
    data = json.load(f)
```

```python
import numpy as np

a = np.random.random_integers(1, 100000, size=20)
a
```

```python
import networkx as nx
import matplotlib.pyplot as plt

for num in a:
    curG = nx.Graph()
    edges = data[num].get("edges", [])
    nodes = data[num].get("nodes", [])
    curG.add_nodes_from(nodes)
    curG.add_edges_from(edges)
    pos = nx.spring_layout(G=curG, method="force", seed=1543)

    plt.figure(figsize=(7, 7))
    nx.draw(curG, pos, node_size=10, width=0.5, with_labels=False)
    plt.title(data[num].get("name"))
    plt.suptitle(data[num].get("description"))
    plt.show()
```
